```python
import whisper
import json

audio_file = "Eveil Télé ⧸ Aucune barrière ： Traduction en nouchi (rire) [nFII73cG57w].wav"

model = whisper.load_model("large-v3")

result = model.transcribe(
    audio_file,
    language="fr",
    task="transcribe",
    verbose=True
)

with open("RJ_001.json", "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)
```

```python
import pympi
import json

with open("RJ_001.json", "r", encoding="utf-8") as f:
    result = json.load(f)

eaf = pympi.Elan.Eaf()
eaf.add_tier("transcription")

for segment in result["segments"]:
    start = int(segment["start"] * 1000)
    end = int(segment["end"] * 1000)
    text = segment["text"].strip()

    if text:
        eaf.add_annotation(
            "transcription",
            start,
            end,
            value=text
        )

eaf.to_file("RJ_001.eaf")
```

```python
import pympi
import json
import os

json_file = "RJ_001.json"
wav_file = "RJ_001.wav"
eaf_file = "RJ_001.eaf"

with open(json_file, "r", encoding="utf-8") as f:
    result = json.load(f)

# ELAN 파일 생성
eaf = pympi.Elan.Eaf()

# WAV 연결
eaf.add_linked_file(wav_file)

# 전사 Tier
eaf.add_tier("transcription")

# Whisper timestamp + transcription
for segment in result["segments"]:
    start = int(segment["start"] * 1000)
    end = int(segment["end"] * 1000)
    text = segment["text"].strip()

    if text:
        eaf.add_annotation(
            "transcription",
            start,
            end,
            value=text
        )

# 저장
eaf.to_file(eaf_file)

print("EAF 생성 완료:", eaf_file)
```
