```python
import whisper
import os
import json

# audio_file = "GabMorrison - Dans le quartier Adjamé avec Lesky (Côte dIvoire).wav"
# uploaded 변수에서 실제 업로드된 파일 이름을 가져옵니다.
audio_file = "GabMorrison - Dans le quartier Adjamé avec Lesky (Côte dIvoire).wav"

# Whisper 모델 선택
model = whisper.load_model("large-v3")

# 전사
result = model.transcribe(
    audio_file,
    language="fr",
    task="transcribe",
    verbose=True
)

# JSON 저장
json_file = "RJ_001.json"

with open(json_file, "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

print(f"Saved: {json_file}")
```
